```python
#category performance
category_performance = (
    clean_df
    .groupby("medicine_category")
    .agg(
        revenue=("revenue_lkr", "sum"),
        units_sold=("quantity", "sum"),
        transactions=("transaction_id", "nunique"),
        medicines=("medicine_id", "nunique")
    )
    .reset_index()
    .sort_values("revenue", ascending=False)
)

category_performance
```

```python
#compare revenue rank vs unit-sales rank. 
#helps separate categories that make lots of money because they're expensive from categories with genuinely high demand.
category_performance["revenue_rank"] = (
    category_performance["revenue"]
    .rank(ascending=False, method="dense")
    .astype(int)
)

category_performance["units_rank"] = (
    category_performance["units_sold"]
    .rank(ascending=False, method="dense")
    .astype(int)
)

category_performance[
    [
        "medicine_category",
        "revenue",
        "units_sold",
        "revenue_rank",
        "units_rank"
    ]
].sort_values("revenue_rank")
```

```python
revenue_median = category_performance["revenue"].median()
units_median = category_performance["units_sold"].median()

def classify_category(row):
    if row["revenue"] >= revenue_median and row["units_sold"] >= units_median:
        return "Core Category"
    
    elif row["revenue"] >= revenue_median and row["units_sold"] < units_median:
        return "High Value"
    
    elif row["revenue"] < revenue_median and row["units_sold"] >= units_median:
        return "High Demand, Lower Value"
    
    else:
        return "Lower Priority"

category_performance["category_class"] = (
    category_performance.apply(classify_category, axis=1)
)

category_performance[
    [
        "medicine_category",
        "revenue",
        "units_sold",
        "category_class"
    ]
].sort_values(
    ["category_class", "revenue"],
    ascending=[True, False]
)
```

```python
medicine_performance = (
    clean_df
    .groupby(["medicine_id", "medicine_name", "medicine_category"])
    .agg(
        revenue=("revenue_lkr", "sum"),
        units_sold=("quantity", "sum"),
        transactions=("transaction_id", "nunique"),
        avg_unit_price=("unit_price_lkr", "mean")
    )
    .reset_index()
)


#contribution to total sales
medicine_performance["revenue_share_pct"] = (
    medicine_performance["revenue"]
    / medicine_performance["revenue"].sum()
    * 100
).round(2)

medicine_performance["unit_share_pct"] = (
    medicine_performance["units_sold"]
    / medicine_performance["units_sold"].sum()
    * 100
).round(2)

medicine_performance.sort_values(
    "revenue",
    ascending=False
).head(10)
```

```python
contribution_analysis = medicine_performance.sort_values(
    "revenue",
    ascending=False
).copy()

contribution_analysis["cumulative_revenue_pct"] = (
    contribution_analysis["revenue"].cumsum()
    / contribution_analysis["revenue"].sum()
    * 100
)

def classify_contribution(cumulative_pct):
    if cumulative_pct <= 80:
        return "top"
    elif cumulative_pct <= 95:
        return "middle"
    else:
        return "bottom"

contribution_analysis["contribution_class"] = (
    contribution_analysis["cumulative_revenue_pct"]
    .apply(classify_contribution)
)


contribution_summary = (
    contribution_analysis
    .groupby("contribution_class")
    .agg(
        medicines=("medicine_id", "count"),
        revenue=("revenue", "sum"),
        units_sold=("units_sold", "sum")
    )
    .reset_index()
)

contribution_summary["revenue_share_pct"] = (
    contribution_summary["revenue"]
    / contribution_summary["revenue"].sum()
    * 100
).round(2)

contribution_summary
```

```python
contribution_summary["medicine_share_pct"] = (
    contribution_summary["medicines"]
    / contribution_summary["medicines"].sum()
    * 100
).round(2)

contribution_summary
```

```python
#Fast vs slow movers
medicine_velocity = (
    clean_df
    .groupby(["medicine_id", "medicine_name"])
    .agg(
        total_units=("quantity", "sum"),
        months_sold=("year_month", "nunique")
    )
    .reset_index()
)

medicine_velocity["avg_units_per_month"] = (
    medicine_velocity["total_units"]
    / medicine_velocity["months_sold"]
).round(2)

medicine_velocity.head()
```

```python
medicine_velocity["avg_units_per_month"].describe().round(2)


```

```python
q1 = medicine_velocity["avg_units_per_month"].quantile(0.25)
q3 = medicine_velocity["avg_units_per_month"].quantile(0.75)

def classify_demand(velocity):
    if velocity >= q3:
        return "High Demand"
    elif velocity <= q1:
        return "Low Demand"
    else:
        return "Moderate Demand"

medicine_velocity["demand_class"] = (
    medicine_velocity["avg_units_per_month"]
    .apply(classify_demand)
)

medicine_velocity["demand_class"].value_counts()

#combine with contribution
medicine_analysis = contribution_analysis.merge(
    medicine_velocity[
        ["medicine_id", "avg_units_per_month", "demand_class"]
    ],
    on="medicine_id",
    how="left"
)

pd.crosstab(
    medicine_analysis["contribution_class"],
    medicine_analysis["demand_class"]
)
```

```python
#Branch performance
branch_performance = (
    clean_df
    .groupby(["branch_id", "branch_name", "district", "province"])
    .agg(
        revenue=("revenue_lkr", "sum"),
        units_sold=("quantity", "sum"),
        transactions=("transaction_id", "nunique"),
        medicines_sold=("medicine_id", "nunique")
    )
    .reset_index()
)

branch_performance["avg_transaction_value"] = (
    branch_performance["revenue"]
    / branch_performance["transactions"]
).round(2)

branch_performance["revenue_share_pct"] = (
    branch_performance["revenue"]
    / branch_performance["revenue"].sum()
    * 100
).round(2)

branch_performance.sort_values(
    "revenue",
    ascending=False
)
```

```python
#branch growth


branch_yoy = (
    clean_df[clean_df["month"] <= 6]
    .groupby(["branch_id", "branch_name", "year"])
    .agg(
        revenue=("revenue_lkr", "sum"),
        units_sold=("quantity", "sum"),
        transactions=("transaction_id", "nunique")
    )
    .reset_index()
)

branch_growth = branch_yoy.pivot(
    index=["branch_id", "branch_name"],
    columns="year",
    values="revenue"
).reset_index()

branch_growth["revenue_growth_pct"] = (
    (branch_growth[2023] - branch_growth[2022])
    / branch_growth[2022]
    * 100
).round(2)

branch_growth.sort_values(
    "revenue_growth_pct",
    ascending=False
)
```

```python
#supplier performance

supplier_performance = (
    clean_df
    .groupby("supplier_name")
    .agg(
        revenue=("revenue_lkr", "sum"),
        units_sold=("quantity", "sum"),
        transactions=("transaction_id", "nunique"),
        medicines_supplied=("medicine_id", "nunique")
    )
    .reset_index()
)

supplier_performance["revenue_share_pct"] = (
    supplier_performance["revenue"]
    / supplier_performance["revenue"].sum()
    * 100
).round(2)

supplier_performance.sort_values(
    "revenue",
    ascending=False
)
```

```python
#Supplier contribution
supplier_contribution = (
    supplier_performance
    .sort_values("revenue", ascending=False)
    .copy()
)

supplier_contribution["cumulative_revenue_pct"] = (
    supplier_contribution["revenue"].cumsum()
    / supplier_contribution["revenue"].sum()
    * 100
).round(2)

supplier_contribution[
    [
        "supplier_name",
        "medicines_supplied",
        "revenue",
        "revenue_share_pct",
        "cumulative_revenue_pct"
    ]
]

top_3_supplier_share = (
    supplier_contribution
    .head(3)["revenue"]
    .sum()
    / supplier_contribution["revenue"].sum()
    * 100
)

print(f"Top 3 suppliers account for {top_3_supplier_share:.2f}% of revenue")
```

```python
#Discount analysis

def classify_discount(rate):
    if rate == 0:
        return "No Discount"
    elif rate <= 0.05:
        return "Low (1-5%)"
    elif rate <= 0.10:
        return "Medium (6-10%)"
    else:
        return "High (>10%)"

clean_df["discount_group"] = (
    clean_df["discount_rate"]
    .apply(classify_discount)
)
```

```python
#summarise performance:
discount_analysis = (
    clean_df
    .groupby("discount_group")
    .agg(
        revenue=("revenue_lkr", "sum"),
        units_sold=("quantity", "sum"),
        transactions=("transaction_id", "nunique"),
        avg_quantity=("quantity", "mean")
    )
    .reset_index()
)

discount_analysis["revenue_share_pct"] = (
    discount_analysis["revenue"]
    / discount_analysis["revenue"].sum()
    * 100
).round(2)

discount_analysis.round(2)
```

```python
#Monthly seasonality by category

category_monthly = (
    clean_df
    .groupby(["medicine_category", "month", "month_name"])
    .agg(
        units_sold=("quantity", "sum"),
        revenue=("revenue_lkr", "sum")
    )
    .reset_index()
)

category_monthly.head()
```

```python
#average monthly units
category_avg = (
    category_monthly
    .groupby("medicine_category")["units_sold"]
    .mean()
    .rename("avg_monthly_units")
)

category_monthly = category_monthly.merge(
    category_avg,
    on="medicine_category"
)

category_monthly["demand_index"] = (
    category_monthly["units_sold"]
    /category_monthly["avg_monthly_units"]
    * 100
).round(1)

peak_months = (
    category_monthly
    .loc[
        category_monthly
        .groupby("medicine_category")["demand_index"]
        .idxmax()
    ]
    [["medicine_category", "month_name", "demand_index"]]
    .sort_values("demand_index", ascending=False)
)

peak_months
```

```python
def classify_priority(row):
    contribution = row["contribution_class"]
    demand = row["demand_class"]

    if contribution == "top" and demand == "High Demand":
        return "Priority Stock"

    elif contribution == "top":
        return "High Value"

    elif demand == "High Demand":
        return "High Demand"

    elif contribution == "bottom" and demand == "Low Demand":
        return "Review"

    else:
        return "Standard"

medicine_analysis["management_priority"] = (
    medicine_analysis.apply(classify_priority, axis=1)
)
```
